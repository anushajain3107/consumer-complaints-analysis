```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Visual styling
plt.style.use('ggplot')
sns.set_theme(style='whitegrid')
pd.set_option('display.max_columns', None)

print('Libraries imported successfully!')
```

## 2. Load Dataset
Loads dataset from `data/consumer_complaints.csv` (or sample file if working on a lightweight environment).

```python
data_path = os.path.join('data', 'consumer_complaints.csv')
sample_path = os.path.join('data', 'sample_consumer_complaints.csv')

target_file = data_path if os.path.exists(data_path) else sample_path
print(f'Loading data from: {target_file}')

df = pd.read_csv(target_file, low_memory=False)
print(f'Total Rows: {len(df):,}, Columns: {df.shape[1]}')
df.head()
```

## 3. Dataset Summary & Inspection
Examine column data types, basic statistics, and null value counts.

```python
# Basic structure
df.info()
```

```python
# Missing value inspection
missing = df.isnull().sum()
missing_pct = (missing / len(df)) * 100
missing_df = pd.DataFrame({'Missing Values': missing, 'Percentage (%)': missing_pct.round(2)})
missing_df[missing_df['Missing Values'] > 0].sort_values(by='Missing Values', ascending=False)
```

## 4. Exploratory Data Analysis (EDA)
Analyze key distributions: Products, Issues, Companies, and Submission Methods.

```python
# Top 10 Products by complaint count
plt.figure(figsize=(10, 5))
top_products = df['product'].value_counts().head(10)
sns.barplot(x=top_products.values, y=top_products.index, palette='Blues_r')
plt.title('Top 10 Complained Products', fontsize=14, fontweight='bold')
plt.xlabel('Number of Complaints')
plt.ylabel('Product')
plt.show()
```

```python
# Top 10 Companies receiving complaints
plt.figure(figsize=(10, 5))
top_companies = df['company'].value_counts().head(10)
sns.barplot(x=top_companies.values, y=top_companies.index, palette='viridis')
plt.title('Top 10 Companies by Complaint Volume', fontsize=14, fontweight='bold')
plt.xlabel('Complaint Count')
plt.ylabel('Company')
plt.show()
```

## 5. Data Cleaning & Feature Engineering
Process dates, calculate response times, and handle missing values.

```python
# Convert dates to datetime format
df['date_received'] = pd.to_datetime(df['date_received'], errors='coerce')
df['date_sent_to_company'] = pd.to_datetime(df['date_sent_to_company'], errors='coerce')

# Compute processing lag
df['response_time_days'] = (df['date_sent_to_company'] - df['date_received']).dt.days
print('Feature engineering completed successfully!')
```

## 6. Text Processing & NLP (Complaint Narratives)
Analyze consumer complaint narratives.

```python
# Filter complaints that include a text narrative
narratives = df[df['consumer_complaint_narrative'].notnull()].copy()
print(f'Complaints with narrative: {len(narratives):,} ({len(narratives)/len(df)*100:.2f}%)')
narratives[['product', 'issue', 'consumer_complaint_narrative']].head(3)
```
